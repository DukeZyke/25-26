```python
import re

# Sample text from Alice in Wonderland for Exercise 2.1
alice_text = """Alice was beginning to get very tired of sitting by her sister on the bank, and of having nothing to do. Once or twice she had peeped into the book her sister was reading, but it had no pictures or conversations in it, "and what is the use of a book," thought Alice, "without pictures or conversations?"""
```

```python
# Simple ELIZA with 5 new patterns

eliza_patterns = [
    (r'.*want to know the reasons why.*feeling depressed.*', 
     'Depression can stem from many causes. Have you considered talking to someone about it?'),
    
    (r'.*feeling stressed.*',
     'Stress is tough. What activities help you relax?'),
    
    (r'.*feelings.*crush.*invalidated.*',
     'Your feelings are valid. It\'s hard when others don\'t understand.'),
    
    (r'.*you (don\'t|do not) understand me.*',
     'I\'m sorry I don\'t understand. Can you tell me more?'),
    
    (r'.*can\'t|cannot focus on.*studies.*',
     'Focus problems are common. Have you tried breaking tasks into smaller parts?')
]

def find_response(user_input):
    """Find response for user input using regex patterns"""
    user_input = user_input.lower()
    
    for pattern, response in eliza_patterns:
        if re.match(pattern, user_input):
            return response
    
    return 'Tell me more about that.'

test_inputs = [
    "I want to know the reasons why I am feeling depressed all the time.",
    "I am feeling stressed.",
    "My feelings towards my crush are invalidated.",
    "You don't understand me.",
    "I can't focus on my studies."
]

print("ELIZA Pattern Responses:")
print("=" * 50)
for user_input in test_inputs:
    response = find_response(user_input)
    print(f"User: {user_input}")
    print(f"ELIZA: {response}")
    print()
```

```python
# ELIZA with repeated question detection

class ElizaWithMemory:
    def __init__(self):
        self.last_question = None
        self.repeated_count = 0
    
    def get_response(self, user_input):
        user_input_lower = user_input.lower()
        
        if self.last_question and user_input_lower == self.last_question:
            self.repeated_count += 1
            sarcastic_responses = [
                "Didn't I just answer that?",
                "Ask me something new, will you?",
                "Same question, same answer!",
                "Are we going in circles here?"
            ]
            return sarcastic_responses[self.repeated_count % len(sarcastic_responses)]
        
        self.last_question = user_input_lower
        self.repeated_count = 0
        
        for pattern, response in eliza_patterns:
            if re.match(pattern, user_input_lower):
                return response
        
        return 'Tell me more about that.'

print("ELIZA With Repeated Question Detection (BONUS):")
print("=" * 50)
eliza = ElizaWithMemory()

conversation = [
    "I am feeling stressed.",
    "I am feeling stressed.",
    "I am feeling stressed.",
    "I can't focus on my studies."
]

for user_input in conversation:
    response = eliza.get_response(user_input)
    print(f"User: {user_input}")
    print(f"ELIZA: {response}")
    print()
```

```python
# Exercise 2.1: Extract words starting with uppercase letter

pattern = r'\b[A-Z]\w*'
uppercase_words = re.findall(pattern, alice_text)

print("Exercise 2.1: Extract Uppercase Words")
print("=" * 50)
print(f"RegEx Pattern: {pattern}")
print(f"\nText: {alice_text}")
print(f"\nWords starting with uppercase:")
print(uppercase_words)
print(f"\nTotal found: {len(uppercase_words)}")
```

```python
# Exercise 2.2: Extract all instances of 'whale' from Moby Dick

with open('melville-moby_dick.md', 'r', encoding='utf-8') as f:
    moby_dick_text = f.read()

pattern = r'(?i)\bwhale\b'
whale_instances = re.findall(pattern, moby_dick_text)

print("Exercise 2.2: Extract 'whale' Instances")
print("=" * 50)
print(f"RegEx Pattern: {pattern}")
print(f"\nTotal instances of 'whale' found: {len(whale_instances)}")
print(f"\nFirst 10 instances:")
for i, whale in enumerate(whale_instances[:10]):
    print(f"  {i+1}. {whale}")
print("  ...")
```

```python
# Exercise 2.3: Extract Jack Sparrow lines from pirates.txt

from nltk.corpus import webtext

pirates_text = webtext.raw('pirates.txt')
pattern = r'(?i).*Jack Sparrow.*'
jack_sparrow_lines = re.findall(pattern, pirates_text)

print("Exercise 2.3: Extract Jack Sparrow Lines")
print("=" * 50)
print(f"RegEx Pattern: {pattern}")
print(f"\nTotal lines with Jack Sparrow found: {len(jack_sparrow_lines)}")
print(f"\nFirst 5 lines:")
for i, line in enumerate(jack_sparrow_lines[:5]):
    print(f"  {i+1}. {line}")
print("  ...")
```

## Summary of RegEx Patterns Used

| Exercise | Pattern | Description |
|----------|---------|-------------|
| 2.1 | `\b[A-Z]\w*` | Words starting with uppercase letter |
| 2.2 | `(?i)\bwhale\b` | Case-insensitive 'whale' word boundary |
| 2.3 | `(?i).*Jack Sparrow.*` | Lines containing Jack Sparrow |
| ELIZA | Multiple patterns | See code above for depression, stress, etc. |
